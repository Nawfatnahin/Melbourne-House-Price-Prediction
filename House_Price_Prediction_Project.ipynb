```python
# Importing essential libraries
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
# Geeting the dataframe
data = pd.read_csv('melb_data.csv')
data
```

```python
data.info() # Looking at info for missing values
```

```python
# Seeing null value percentage
data.isnull().mean()*100
```

```python
# Changing some features 
data['YearBuilt'] = 2026 - data['YearBuilt']

```

```python
# Creating input and output 
x = data.drop(columns =['Regionname','Propertycount','Price','Postcode','Address','Date'], axis =1 )
y = data["Price"]
```

```python
# Train and Test data splitting
from sklearn.model_selection import train_test_split
X_train, X_test , Y_train , Y_test = train_test_split(x,y,test_size= 0.2, random_state=3)
```

```python
X_train
```

```python
# Importing for data proocessing 
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
```

```python
# Listing columns for different processing 
missing_imput_cols1 = ['YearBuilt','BuildingArea']
missing_imput_cols2 = ['Car']
ohe_cols = ['Suburb','Method','Type','SellerG']

# Making pipeline to process a column step by step 
pipeline = Pipeline([
   ('filling', SimpleImputer(strategy="most_frequent")),
    ('encoding', OneHotEncoder(handle_unknown='ignore'))
    ])

# Making transformer to process all the columns in one go
trf = ColumnTransformer(transformers=[
    ('trf1',SimpleImputer(strategy= 'constant', fill_value=0),missing_imput_cols1),
    ('trf2',SimpleImputer(strategy="most_frequent"),missing_imput_cols2),
    ('trf3',OneHotEncoder(handle_unknown='ignore'),ohe_cols),
    ('trf4', pipeline, ['CouncilArea']), 
],remainder ='passthrough')
```

```python
X_train_new = trf.fit_transform(X_train)
X_test_new = trf.transform(X_test)
```

```python
# Scaling data
scaler = StandardScaler(with_mean=False)

X_train_scaled = scaler.fit_transform(X_train_new)
X_test_scaled = scaler.transform(X_test_new)
```

```python
# Now, We will craft the models 
from sklearn.ensemble import RandomForestRegressor
```

```python
lr = RandomForestRegressor()

lr.fit(X_train_scaled,Y_train)
```

```python
lr.score(X_test_scaled,Y_test)
```

```python
# Second Model
from sklearn.tree import DecisionTreeRegressor
```

```python
dtr = DecisionTreeRegressor(min_samples_split=100, max_depth= 90)
```

```python
dtr.fit(X_train_scaled,Y_train)
```

```python
dtr.score(X_test_scaled,Y_test)
```

```python
# Now , we will compare MAE, RMSE and R2 for each model
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

y_pred_lr = lr.predict(X_test_scaled)
y_pred_dtr = dtr.predict(X_test_scaled)

results = [
    {
        "Model": "RandomForestRegressor",
        "MAE": mean_absolute_error(Y_test, y_pred_lr),
        "RMSE": np.sqrt(mean_squared_error(Y_test, y_pred_lr)),
        "R2": r2_score(Y_test, y_pred_lr)
    },
    {
        "Model": "DecisionTreeRegressor",
        "MAE": mean_absolute_error(Y_test, y_pred_dtr),
        "RMSE": np.sqrt(mean_squared_error(Y_test, y_pred_dtr)),
        "R2": r2_score(Y_test, y_pred_dtr)
    },

]

pd.DataFrame(results)

# Seeing the result data frame we can say ->
# 1. Random Forest has lower MAE and RMSE which means smaller average errors.
# 2. Random Forest has higher R² (0.837 vs 0.728) meaning randome forest regression fits better.
# 3. RMSE > MAE for both --> some large errors (outliers) are present pulling RMSE up.
```

```python
# extracting the most influential features

feature_names = trf.get_feature_names_out()


rf_model = lr
importances = rf_model.feature_importances_


importance_df = pd.DataFrame({
    'Feature': feature_names,
    'Importance': importances
}).sort_values(by='Importance', ascending=False)

top_10 = importance_df.head(10)

# 4. Plot horizontal bar chart
plt.figure(figsize=(12, 8))
sns.barplot(
    data=top_10,
    x='Importance',
    y='Feature',
    palette='Blues_r'
)
plt.title('Top 10 Most Influential Features on Melbourne House Prices', fontsize=14, weight='bold', pad=15)
plt.xlabel('Relative Feature Importance Score', fontsize=12)
plt.ylabel('Feature', fontsize=12)
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()

```
